```python
dataset_dir = Path.home() / 'userdata' / 'open-score-string-quartets'
dataset_data_dir = dataset_dir / 'data'
dataset_score_dir = dataset_dir / 'scores'

dataset_dir.exists()
```

```python
metadata = load_ossq_metadata(dataset_data_dir / 'scores.yaml')
```

## Compare System order: Synthetic vs. Scanned

```python
score_indices = [
  '7358579',
]
```

```python
IDX = 0

score_id = score_indices[IDX]
score_dir = dataset_score_dir / metadata[score_id]['path']
synth_systems = sorted((score_dir / 'images' / 'synthetic' / 'cropped').glob(f'sq{score_id}*.png'))
scanned_systems = sorted((score_dir / 'images' / 'scanned' / 'cropped').glob(f'sq{score_id}*.png'))

print(len(synth_systems), len(scanned_systems))

data_list = list(zip(synth_systems, scanned_systems))

ImageComparePreview(data_list).show()
```

## Compare Image-LMXE Alignment: Synthetic

```python
random.seed(260303)

pdf_type = 'synthetic'
n = 2

data_list = []

for _, body in metadata.items():
  sqid = body['sqid']
  score_dir = dataset_score_dir / body['path']
  
  img_paths = (score_dir / 'images' / pdf_type / 'systemwise').glob(f'{sqid}*.png')
  img_paths = sorted(img_paths)
  
  sampled = random.sample(img_paths, min(n, len(img_paths)))
  data_list += [
    (p, score_dir / 'lmxe' / pdf_type / 'systemwise' / p.name.replace('.png', '.lmxe'))
    for p in sampled
  ]

len(data_list)
```

```python
IDX = 9
LMXEPreview(data_list, index=IDX).show()
```

## Compare Image-LMXE Alignment: Scanned

```python
random.seed(260303)

pdf_type = 'scanned'
n = 2

data_list = []

for _, body in metadata.items():
  sqid = body['sqid']
  score_dir = dataset_score_dir / body['path']
  
  img_paths = (score_dir / 'images' / pdf_type / 'systemwise').glob(f'{sqid}*.png')
  img_paths = sorted(img_paths)
  
  if len(img_paths) < 1:
    continue
  
  sampled = random.sample(img_paths, min(n, len(img_paths)))
  data_list += [
    (p, score_dir / 'lmxe' / pdf_type / 'systemwise' / p.name.replace('.png', '.lmxe'))
    for p in sampled
  ]

len(data_list)
```

```python
IDX = 0
LMXEPreview(data_list, index=IDX).show()
```
