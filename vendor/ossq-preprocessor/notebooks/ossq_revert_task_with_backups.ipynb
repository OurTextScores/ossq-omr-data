```python
dataset_dir = Path.home() / 'userdata' / 'open-score-string-quartets'
dataset_data_dir = dataset_dir / 'data'
dataset_score_dir = dataset_dir / 'scores'

dataset_dir.exists()
```

```python
metadata = load_ossq_metadata(dataset_data_dir / 'scores.yaml')
```

## revert work with backed up files

```python
pdf_type = 'synthetic'
file_type = 'abc'
level = 'systemwise'
sub_dir = 'uncleaned'

for score_id, body in metadata.items():
  sqid = body['sqid']
  sq_dir = dataset_score_dir / body['path']
  file_dir = sq_dir / file_type / pdf_type / level
  
  for f in file_dir.glob(f'{sqid}*.{file_type}'):
    f.unlink()
  
  for f in (file_dir / sub_dir).glob(f'{sqid}*.{file_type}'):
    shutil.copy(
      f,
      file_dir / f.name
    )
```
