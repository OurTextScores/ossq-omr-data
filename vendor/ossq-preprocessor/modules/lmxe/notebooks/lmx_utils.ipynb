```python
def split_tokens_by_key_token(token_list, key_token, include_key=False, compare_fn=eq):
  groups = []
  
  cur_group = []
  
  for tok in token_list:
    if compare_fn(tok, key_token):
      if len(cur_group) > 0:
        groups.append(cur_group)
        cur_group = []
      
      if include_key:
        groups.append([tok])
      
      continue
    
    cur_group.append(tok)
  
  else:
    groups.append(cur_group)
  
  return groups
```

```python
def get_new_system_element():
  el = ET.Element('print')
  el.set('new-system', 'yes')
  
  staff_layout = ET.SubElement(el, 'staff-layout', attrib={'number': "1"})
  
  return el
```

## Proto: Parse MusicXML to SQLMXE
* SQLMXE: combined measure-wise String Quartet LMX by page

```python
# sub_dir, mscore_id = itemgetter('path', 'mscore_id')(mscore_data[0])
# mscore_id = '7362818'
mscore_id = '7300376'
sub_dir = score_dir / score_metadata[mscore_id]['path']
mscore_path = score_dir / sub_dir / f'sq{mscore_id}.musicxml'

mscore_path.exists()
```

```python
with open(mscore_path, 'r') as f:
  mxml = f.read()

xml_tree = ET.ElementTree(ET.fromstring(mxml))
```

```python
part_list = xml_tree.getroot().find('part-list')
for part in part_list.findall('score-part'):
  print(part.find("score-instrument/instrument-name").text)
```

```python
mxl = MxlFile(xml_tree)

mxl.get_string_quartet_parts()
```

```python
mxl = MxlFile(xml_tree)

try:
  sq_parts = mxl.get_string_quartet_parts()
except:
  sq_parts = mxl.tree.find("part")

if len(sq_parts) != 4 or any([p.tag != "part" for p in sq_parts]):
  print("No <part> element found.", file=sys.stderr)
  exit()

sq_linearizer = SQLinearizer(
  errout=sys.stderr
)
lmx_parts = sq_linearizer.process_multi_parts(sq_parts)
```

```python
print(lmx_parts[0])
print(lmx_parts[1])
```

```python
pages_by_parts = [
  split_tokens_by_key_token(part, "<new-page>")
  for part in lmx_parts
]

# check number of pages in each part
all( [ len(pages_by_parts[0]) == len(pages) for pages in pages_by_parts[1:] ] )
```

```python
len(pages_by_parts[0]), pages_by_parts[0][0]
```

```python
systems_by_parts = [
  [
    split_tokens_by_key_token(page, "<new-system>", include_key=False)
    for page in part
  ]
  for part in pages_by_parts
]

# check number of systems in each page
all([
  all( [len(pages[0]) == len(part) for part in pages[1:]] )
  for pages in zip(*systems_by_parts)
])
```

```python
measures_by_parts = [
  [
    [
      split_tokens_by_key_token(system, "measure", include_key=False)
      for system in pages
    ]
    for pages in part
  ]
  for part in systems_by_parts
]

# check number of measures in each system
all([
  all([
    all([
      len(systems[0]) == len(measure)
      for measure in systems[1:]
    ])
    for systems in zip(*pages)
  ])
  for pages in zip(*measures_by_parts)
])
```

```python
for p_i, pages in enumerate(zip(*measures_by_parts)):
  for s_i, system in enumerate(zip(*pages)):
    for part_idx, measures in enumerate(system[1:]):
      if len(system[0]) != len(measures):
        print(part_idx+1, p_i+1, s_i+1, len(system[0]), len(measures))
```

```python
lmx_score = []

for pages in zip(*measures_by_parts):
  lmx_page = []
  
  for systems in zip(*pages):
    lmx_system = []
    
    for measure in zip(*systems):
      lmx_measure = ['measure']
      
      for p_i, part in enumerate(measure):
        lmx_measure += [f'part:{p_i+1}'] + part
      
      lmx_measure_str = ' '.join(lmx_measure)
      lmx_system.append(lmx_measure_str)
    
    lmx_page.append(lmx_system)
  
  lmx_score.append(lmx_page)
```

## Convert openScore SQ Corpus to SQLMXE

```python
from lmx_utils import linearize_lmxe

for item in tqdm(mscore_data, desc='score'):
  sub_dir, mscore_id = itemgetter('path', 'mscore_id')(item)
  xml_path = score_dir / sub_dir / f'sq{mscore_id}.musicxml'
  
  lmxe_dir = score_dir / sub_dir / 'lmxe'
  lmxe_dir.mkdir(exist_ok=True)
  
  with open(xml_path, 'r') as f:
    xml_str = f.read()
  
  lmx_score = linearize_lmxe(xml_str)
  lmx_score
  
  for p_i, page in enumerate(tqdm(lmx_score, desc='page', leave=False)):
    lmxe_str = '\n\n'.join([ '\n'.join(system)  for system in lmx_score[0] ])
    lmxe_path = lmxe_dir / f'sq{mscore_id}:{str(p_i).zfill(4)}.lmxe'
    
    with open(lmxe_path, 'w') as f:
      f.write(lmxe_str)
```

## Convert system-wise LMX to Music XML

```python
# sub_dir, mscore_id = itemgetter('path', 'mscore_id')(mscore_data[0])
# mscore_id = '7300376'
mscore_id = '7362818'
sub_dir = score_dir / score_metadata[mscore_id]['path']
lmxe_dir = score_dir / sub_dir / 'lmxe'

lmxe_dir.exists()
```

```python
lmxe_paths = list(sorted(lmxe_dir.glob('*.lmxe')))

all([ l_p.exists() for l_p in lmxe_paths ])
```

```python
lmxe_path = lmxe_paths[0]
```

```python
with open(lmxe_path, 'r') as f:
  lmxe_str = f.read()

lmxe_tokens_by_systems = [ system.replace('\n', ' ').split(' ') for system in lmxe_str.split('\n\n') ]

lmxe_tokens_by_systems
```

```python
lmxe_measures_by_systems = [
  split_tokens_by_key_token(system, "measure", include_key=False)
  for system in lmxe_tokens_by_systems
]

dprint(lmxe_measures_by_systems[0])
```

```python
lmxe_parts_by_measures = [
  [ 
    split_tokens_by_key_token(measure, f'part:', include_key=False, compare_fn=contains)
    for measure in system
  ]
  for system in lmxe_measures_by_systems
]

# print(len(lmxe_parts_by_measures[0][0][0]))
# dprint(lmxe_parts_by_measures[0])

lmxe_systems_by_parts = [ [] for _ in range(4) ]

for system in lmxe_parts_by_measures:
  for p_i, part in enumerate(zip(*system)):
    measuers = [ ' '.join(['measure'] + measure) for measure in part ]
    lmxe_systems_by_parts[p_i].append(
      ' '.join(measuers)
    )

dprint(list(zip(*lmxe_systems_by_parts))[0])
```

```python
xml_parts = []

for part in lmxe_systems_by_parts:
  part_xml_with_system_breaks = None
  for s_i, system in enumerate(part):
    part_xml = Delinearizer( errout=sys.stderr ).process_text( system )
    
    if s_i > 0:
      part_xml.find('measure').insert(0, get_new_system_element())
      part_xml_with_system_breaks.extend(part_xml.findall('measure'))
    
    else:
      part_xml_with_system_breaks = part_xml
    
  # print(part_xml_with_system_breaks.findall('.//measure/print[@new-system="yes"]'))
  
  xml_parts.append(part_xml_with_system_breaks)

len(xml_parts)
```

```python
score_etree = string_quartet_parts_to_score(xml_parts)

output_xml = str(
  ET.tostring(
    score_etree.getroot(),
    encoding="utf-8",
    xml_declaration=True
  ), 
  "utf-8"
)
```
